## Category-wise Analysis:

```python
category_analysis = df.groupby('Category')[['Sales', 'Profit']].sum()

category_analysis
```

### Category-wise Findings

1. Technology generated the highest total sales (836,154.03) and the highest total profit (145,454.95).

2. Furniture generated 741,999.80 in sales but only 18,451.27 in profit, showing a relatively small profit compared with its sales.

3. Office Supplies generated lower sales than Furniture but significantly higher profit (122,490.80 compared with 18,451.27).

## Region-wise Analysis:

```python
region_analysis = df.groupby('Region')[['Sales', 'Profit']].sum()

region_analysis
```

### Region-wise Findings

2. The South region recorded the lowest sales (391,721.91).

## Sub Category-wise Profit Analysis:

```python
subcategory_analysis = df.groupby('Sub-Category')[['Sales', 'Profit']].sum()

subcategory_analysis = subcategory_analysis.sort_values(
    by='Profit',
    ascending=False
)

subcategory_analysis
```

### Sub-Category-wise Findings

1. Copiers generated the highest total profit (55,617.82), even though their sales were lower than Phones and Chairs.

2. Phones recorded the highest sales (330,007.05), but Copiers generated higher profit.

3. Tables recorded the largest total loss (-17,725.48), despite generating 206,965.53 in sales.

4. Bookcases (-3,472.56) and Supplies (-1,189.10) also recorded negative total profits.

5. Out of 17 sub-categories, 3 recorded negative total profits.

## Discount and Profit Analysis:
To investigate does a higher discount correspond to lower profit in this dataset?

```python
discount_analysis = df.groupby('Discount').agg(
    Total_Records=('Profit', 'count'),
    Average_Profit=('Profit', 'mean'),
    Total_Profit=('Profit', 'sum')
)

discount_analysis
```

### Discount and Profit Findings

1. Records with 0% discount generated the highest total profit (320,987.60).

2. The 10% discount level recorded the highest average profit (96.06).

3. All discount levels from 30% to 80% recorded negative average profits.

4. The 50% discount level had the lowest average profit (-310.70).

5. The 70% discount level generated the largest total loss (-40,075.36).

6. The dataset shows an association between higher discount levels and lower profitability. However, this does not establish that discounts alone cause losses.

## Customer Segment Analysis
Now let's analyze the three customer segments:
- Consumer
- Corporate
- Home Office
We want to know which customer segment contributes the most sales and profit.

```python
segment_analysis = df.groupby('Segment').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Average_Sales=('Sales', 'mean'),
    Average_Profit=('Profit', 'mean')
)

segment_analysis
```

### Customer Segment Findings

1. The Consumer segment generated the highest total sales (1,161,401.35) and total profit (134,119.21).

2. The Home Office segment recorded the highest average sales per record (240.97).

3. The Home Office segment also achieved the highest average profit per record (33.82).

4. Although Consumer leads in total profit, Home Office performs better in average profit per record.

## Year-wise Sales and Profit Analysis
To explore how business performance changes over time.

To convert Order date columns to Pandas datetme format:

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

df['Order Date'].dtype
```

To create a new column called Year from Order Date.

```python
df['Year'] = df['Order Date'].dt.year

df[['Order Date', 'Year']].head()
```

```python
yearly_analysis = df.groupby('Year').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Average_Sales=('Sales', 'mean'),
    Average_Profit=('Profit', 'mean')
)

yearly_analysis
```

### Year-wise Findings

1. The year 2017 recorded the highest total sales (733,215.26) and total profit (93,439.27).

2. In 2015, total sales decreased compared with 2014, but total profit increased from 49,543.97 to 61,618.60.

3. The year 2016 achieved the highest average profit per record (31.62).

4. Average profit per record decreased from 31.62 in 2016 to 28.21 in 2017, despite an increase in total profit.

## Count Records per Year:
To investigate why 2017 has higher total profit but lower average profit than 2016.

```python
yearly_records = df.groupby('Year').size()

yearly_records
```

The number of records increased from 1,993 in 2014 to 3,312 in 2017. Although 2017 achieved the highest total profit, its average profit per record was lower than in 2016.

## Correaltion between profits and Sales:

```python
sales_profit_correlation = df['Sales'].corr(df['Profit'])

print("Sales-Profit Correlation:", sales_profit_correlation)
```

### Sales and Profit Correlation

The correlation between Sales and Profit is approximately 0.48, indicating a moderate positive linear relationship. Generally, higher sales tend to be associated with higher profits, but this relationship is not perfect. Some high-sales transactions can still generate losses.
